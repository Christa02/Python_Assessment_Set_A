```python
number_students = int(input())
students = {}
for val in range(number_students):
  student_data = input()
  student_data_list = student_data.split(" ")
  student_mark_average = ((float(student_data_list[1])+float(student_data_list[2])+float(student_data_list[3]))) / 3
  if ((student_mark_average >= 50) and (float(student_data_list[1]) >= 40) and (float(student_data_list[2]) >= 40) and (float(student_data_list[3]) >= 40)):
    if ((student_mark_average >= 80) and (student_mark_average <= 100)):
      students[student_data_list[0]] = (student_mark_average, "Passed", "A")
    elif ((student_mark_average >= 70) and (student_mark_average <= 79)):
      students[student_data_list[0]] = (student_mark_average, "Passed", "B")
    elif ((student_mark_average >= 60) and (student_mark_average <= 69)):
      students[student_data_list[0]] = (student_mark_average, "Passed", "C")
    elif ((student_mark_average >= 50) and (student_mark_average <= 59)):
      students[student_data_list[0]] = (student_mark_average, "Passed", "D")
  else:
    students[student_data_list[0]] = (student_mark_average, "Failed", "")

passed_count = failed_count = top_avg = 0
top_performer = ""
for key, val in students.items():
  if (val[1] == "Passed"):
    print(f"{key} - Average: {val[0]:.2f} - {val[1]} - Grade: {val[2]}")
    passed_count += 1
    if (val[0] > top_avg):
      top_avg = val[0]
      top_performer = key.title()
  else:
    print(f"{key} - Average: {val[0]:.2f} - {val[1]}")
    failed_count += 1

print(f"Passed: {passed_count}")
print(f"Failed: {failed_count}")
print(f"Top Performer: {top_performer}")
```
