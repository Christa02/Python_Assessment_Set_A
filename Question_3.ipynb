```python
class Product:
  def __init__(self, product_id, product_name, price, available_stock):
    self.product_id = product_id
    self.product_name = product_name
    self.price = price
    self.available_stock = available_stock
```

```python
class ShoppingCart(Product):

  def add_product(self, prod_id, prod_qty):
    if ((prod_qty > 0) and (self.available_stock > prod_qty)):
      return "Sufficient Stock"
    else:
      return "Insufficient Stock"

```
