```python
Customer_Name = input()
Units_Consumed = float(input())
service_charge = 50
units_charge = 0
electricity_charge = 0.0
total_bill = 0.0
if (Units_Consumed > 0):
  if (Units_Consumed <= 100):
    units_charge = 2
  elif ((Units_Consumed >= 101) and (Units_Consumed <= 200)):
    units_charge = 3
  elif ((Units_Consumed >= 201) and (Units_Consumed <= 400)):
    units_charge = 5
  else:
    units_charge = 7
  electricity_charge = units_charge * Units_Consumed
  total_bill = electricity_charge + service_charge
  print(f"Customer: {Customer_Name}")
  print(f"Units Consumed: {Units_Consumed:.2f}")
  print(f"Electricity Charge: {electricity_charge:.2f}")
  print(f"Service Charge: {service_charge:.2f}")
  print(f"Total Bill: {total_bill:.2f}")
else:
  print(f"Invalid Units")
```
